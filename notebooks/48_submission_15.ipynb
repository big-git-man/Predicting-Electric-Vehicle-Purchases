# Submission 15

This submission blends my 5-fold Submission 14 predictions with a strong public prediction from the external archive. The blend uses 85% Submission 14 and 15% of the archived 0.94657 prediction.

This is an experimental submission, so the Kaggle score will tell us whether the external prediction adds useful ranking information.

```python
from pathlib import Path
import pandas as pd

ROOT = Path(r'C:\\Users\\aakif\\Documents\\DataCompetition')
S14 = ROOT / 'submissions' / 'submission_14.csv'
EXT = ROOT / 'external_data' / 's6e9_zoom_zoom_baseline' / 'submission_latest_best.csv'
OUT = ROOT / 'submissions' / 'submission_15.csv'

s14 = pd.read_csv(S14)
ext = pd.read_csv(EXT)

if not s14['id'].equals(ext['id']):
    ext = ext.set_index('id').loc[s14['id']].reset_index()

pred = 0.85 * s14['Will_Buy_EV'] + 0.15 * ext['Will_Buy_EV']
submission = pd.DataFrame({'id': s14['id'], 'Will_Buy_EV': pred})
submission.to_csv(OUT, index=False)

print(f'Saved: {OUT}')
print(f'Rows: {len(submission):,}')
```
